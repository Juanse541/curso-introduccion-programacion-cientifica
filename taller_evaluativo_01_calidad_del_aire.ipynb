**Ejercicio 6:** Limpiar la serie horaria

```python
lecturas_pm25 = [8.4, 12.7, -999.0, 15.2, 41.8, 9.6, -999.0, 58.3, 22.0, 13.9, 7.5, 36.1]
lecturas_validas = []
suma_pm25 = 0.0
lecturas_descartadas = 0

for i in range(len(lecturas_pm25)):
  lectura = lecturas_pm25[i]
  if lectura == -999.0:
    lecturas_descartadas += 1
  else:
    lecturas_validas.append(lectura)
    suma_pm25 += lectura

# (preguntar a profe para explicame en linea 9)
if len(lecturas_validas) > 0:
    promedio_pm25 = suma_pm25 / len(lecturas_validas)
else:
    promedio_pm25 = 0.0
print(f"Cantidad lectura Válidas: {len(lecturas_validas)}")
print(f"Cantidad de Descartadas: {lecturas_descartadas}")
print(f"Promedio: {promedio_pm25}")
```

**Ejercicio 7:** Estadística descriptiva de la serie limpia

```python
# preguntar a profe para explicarme mas aclarar es dudo.
# Para evitar errores si lecturas_validas está vacía
if not lecturas_validas:
    maximo_pm25 = None
    minimo_pm25 = None
    rango = None
    desviacion_estandar = None
    print("No hay lecturas válidas para calcular estadísticas.")
else:
    # TODO: inicialice maximo_pm25 y minimo_pm25 con el primer elemento de lecturas_validas
    maximo_pm25 = lecturas_validas[0]
    minimo_pm25 = lecturas_validas[0]

    # TODO: recorra lecturas_validas con un for y, con if, actualice maximo_pm25
    #       cuando encuentre un valor mayor y minimo_pm25 cuando encuentre uno menor
    for lectura in lecturas_validas:
        if lectura > maximo_pm25:
            maximo_pm25 = lectura
        if lectura < minimo_pm25:
            minimo_pm25 = lectura

    # TODO: calcule el rango restando el mínimo al máximo
    rango = maximo_pm25 - minimo_pm25

    # TODO: declare el acumulador suma_desviaciones en 0.0
    suma_desviaciones_cuadradas = 0.0

    # TODO: en un segundo for sobre lecturas_validas, acumule en suma_desviaciones
    #       la diferencia entre cada lectura y promedio_pm25, elevada al cuadrado
    # Se asume que promedio_pm25 ya ha sido calculado en el Ejercicio 6
    for lectura in lecturas_validas:
        suma_desviaciones_cuadradas += (lectura - promedio_pm25) ** 2

    # TODO: calcule desviacion_estandar elevando a la potencia 0.5 el cociente
    #       entre suma_desviaciones y la cantidad de lecturas válidas
    # Para la desviación estándar muestral, se divide por (n-1)
    if len(lecturas_validas) > 1:
        desviacion_estandar = (suma_desviaciones_cuadradas / (len(lecturas_validas) - 1)) ** 0.5
    else:
        desviacion_estandar = 0.0 # Si hay 0 o 1 elemento, la desviación estándar es 0

    # TODO: muestre con print() el máximo, el mínimo, el rango y la desviación
    print(f"Máximo: {maximo_pm25}")
    print(f"Mínimo: {minimo_pm25}")
    print(f"Rango: {rango}")
    print(f"Desviación Estándar: {desviacion_estandar}")
```

**Ejercicio 8:** Decaimiento tras cerrar la fuente de emisión

```python
concentracion = 58.3
factor_decaimiento = 0.85
umbral_buena = 12.0


horas_transcurridas = 0

while concentracion > umbral_buena:
    concentracion *= factor_decaimiento
    horas_transcurridas += 1

print(f"Cuantas horas pasaron: {horas_transcurridas} horas")
print(f"Concentración final: {concentracion:.2f} m3")
```
